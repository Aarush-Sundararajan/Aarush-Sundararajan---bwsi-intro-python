## FizzBuzz Problem

Write a short program that prints each number from 1 to 100 except:

- For each multiple of 3, print "Fizz" instead of the number

- For each multiple of 5, print "Buzz" instead of the number

- For numbers which are multiples of both 3 and 5, print "FizzBuzz" instead of the number

### Q.1 Please outline in words the steps that should be executed for each iteration of the loop

### A.1
For each number from 1 to 100: first check if it is divisible by both 3 and 5, and if so print "FizzBuzz". Otherwise, check if it is divisible by 3 and print "Fizz". Otherwise, check if it is divisible by 5 and print "Buzz". If none of these conditions are met, print the number itself.

### Q.2 Please implement the solution described in Q.1 with comments

```python
# Loop through numbers 1 to 100
for i in range(1, 101):
    # Check if divisible by both 3 and 5 first
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    # Check if divisible by 3
    elif i % 3 == 0:
        print("Fizz")
    # Check if divisible by 5
    elif i % 5 == 0:
        print("Buzz")
    # Otherwise print the number
    else:
        print(i)
```

## Understanding Code Flow

### Q.3 Please complete the commented lines below with descriptions of the code as indicated. 

```python
# Create list to store prime numbers
prime_numbers_list = []

# This for loop iterates over all integers from 2 to 10
for n in range(2,11):
    # Set is_prime as True
    is_prime = True
    # This for loop iterates over all integers from 2 to n-1 (potential divisors)
    for k in range(2, n):
        # This condition checks if n is evenly divisible by k
        if n % k == 0:
            # If n % k == 0, then n is not prime; print its factorization
            print(f"{n} = {k} * {n // k}")
            is_prime = False
            break
    # This condition checks if is_prime is True
    if is_prime:
        # If n is prime, then print as output
        print(f"{n} is a prime number!")
        prime_numbers_list.append(n)

print(f"\nPrime numbers: {prime_numbers_list}")
```

### A.4
